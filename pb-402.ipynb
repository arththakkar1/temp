```python
import requests

API_KEY="a2d8a5383655bf8e313435d0b68ae065"
lat= 28.6517178
lon= 77.2219388
Link=f"https://api.openweathermap.org/data/2.5/weather?lat={lat}&lon={lon}&appid={API_KEY}"

data=requests.get(Link)
res=data.json()

Pres=res["main"]["pressure"]
hum=res["main"]["humidity"]
sea_lev=res["main"]["sea_level"]
Vis=res["visibility"]
Tim=res["timezone"]
Sun=res["sys"]["sunset"]
Des=res["weather"][0]["description"]
Speed=res["wind"]["speed"]

print("Pressure:", Pres, "hPa")
print("Humidity:", hum, "%")
print("Sea Level:", sea_lev, "hPa")
print("Visibility:", Vis, "meters")
print("Timezone:", Tim, "seconds from UTC")
print("Sunset:", Sun)
print("Weather Description:", Des)
print("Wind Speed:", Speed, "m/s")
```
