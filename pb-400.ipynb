```python
from bs4 import BeautifulSoup
import pandas as pd

with open("fakepython.html","r",encoding="utf-8") as file:
    soup=BeautifulSoup(file,"html.parser")

titles=soup.find_all("h2",class_="title")

title_fil=[]

for i in titles:
    i=i.get_text(strip=True)

    if "Python" in i:
        title_fil.append(i)

print(title_fil)

Get_Data=soup.find_all("div",class_="card-content")

print("-----------------")

data={
    "Title":[],
    "Location":[],
    "Company":[]
}

for i in Get_Data:
    title=i.find("h2",class_="title").get_text(strip=True)
    loc=i.find("p",class_="location").get_text(strip=True)
    comp=i.find("h3",class_="company").get_text(strip=True)

    data["Title"].append(title)
    data["Location"].append(loc)
    data["Company"].append(comp)

    print(f"Title:{title}")
    print(f"Location:{loc}")
    print(f"Company:{comp}")

    print("-----------------")

df=pd.DataFrame(data)
```
