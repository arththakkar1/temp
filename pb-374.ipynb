```python
import requests
from bs4 import BeautifulSoup

url = "https://indianexpress.com/"


response = requests.get(url)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

# Try to locate a section whose class/id contains 'top'
top_news_section = None

for tag in soup.find_all(["div", "section"]):
    classes = " ".join(tag.get("class", []))
    tag_id = tag.get("id", "")

    if "top" in classes.lower() or "top" in tag_id.lower():
        top_news_section = tag
        break

if top_news_section:
    print("Top News Headlines:\n")

    headlines = top_news_section.find_all(["h1", "h2", "h3", "h4"])

    for i, headline in enumerate(headlines, start=1):
        text = headline.get_text(strip=True)
        if text:
            print(f"{i}. {text}")
else:
    print("Top News section not found.")
```
